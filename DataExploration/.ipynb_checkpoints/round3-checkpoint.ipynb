# Round 3 Data Exploration

Products: **VELVETFRUIT_EXTRACT**, **HYDROGEL_PACK**, **VEV_XXXX** (derivatives/options)  
Days: 0, 1, 2

```python
import sys, os
sys.path.insert(0, os.path.join(os.path.dirname(os.getcwd()), ""))
from plotter import Plotter

DATA_DIR = "../data/round3"
days = [0, 1, 2]
```

```python
import pandas as pd
```

```python
all_prices = pd.concat(
    [pd.read_csv(f"{DATA_DIR}/prices_round_3_day_{d}.csv", delimiter=";") for d in days],
    ignore_index=True,
)

products = all_prices['product'].unique().tolist()
print(f"total rows: {len(all_prices)}")
print(f"products ({len(products)}): {products}")

per_product = {p: all_prices[all_prices['product'] == p].reset_index(drop=True) for p in products}
for p, df in per_product.items():
    print(f"  {p}: {len(df)} rows")
```

```python
# add spread column to each product frame
for df in per_product.values():
    df['spread'] = df['ask_price_1'] - df['bid_price_1']

velvet = per_product.get('VELVETFRUIT_EXTRACT')
hydrogel = per_product.get('HYDROGEL_PACK')
vev_products = sorted([p for p in products if p.startswith('VEV_')])
print(f"VEV strikes: {vev_products}")
```

```python
# Load trades with day column (trades CSVs don't have one, so we add it)
all_trades = pd.concat(
    [pd.read_csv(f"{DATA_DIR}/trades_round_3_day_{d}.csv", delimiter=";")
       .assign(day=d)
     for d in days],
    ignore_index=True,
)

# Merge price columns onto trades at matching (day, timestamp, product)
all_trades = all_trades.rename(columns={"symbol": "product"})
all_trades = all_trades.merge(all_prices, on=["day", "timestamp", "product"], how="left")

trades_by_product = {p: all_trades[all_trades["product"] == p].reset_index(drop=True) for p in products}
for p, df in trades_by_product.items():
    print(f"  {p}: {len(df)} trades")
```

```python
# spread distribution for the main tradable products
for name in ['VELVETFRUIT_EXTRACT', 'HYDROGEL_PACK']:
    df = per_product.get(name)
    if df is None:
        continue
    print(f"\n{name} spread counts:")
    print(df['spread'].value_counts().sort_index().head(20))
```
